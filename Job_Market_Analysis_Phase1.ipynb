## 1. Project Overview

This project focuses on analysing job-posting data from India and the United States to understand patterns in job opportunities, skill requirements, advertised salaries, experience requirements, and remote-work arrangements.

The project uses a public job-postings dataset containing information about job titles, roles, companies, locations, seniority levels, skills, and annual salary values where available.

Using Microsoft Excel, SQL, Python, and Power BI, the project will involve data cleaning, exploratory data analysis, skill-frequency analysis, salary analysis, and dashboard development.

The main aim is to transform raw job-posting data into meaningful insights that can help students, job seekers, and career planners better understand the skills and requirements mentioned in job advertisements.

## 2. Business Problem

Students and job seekers often find it difficult to identify which technical and professional skills are frequently mentioned in job advertisements, what experience levels employers request, and how advertised salaries vary across roles and locations.

Although job advertisements contain useful information about employer requirements, the information can be scattered across different roles and locations. Without systematic analysis, it may be difficult to identify patterns and make informed career-planning decisions.

This project addresses the problem by analysing job-posting data from India and the United States to identify frequently mentioned skills, examine available salary patterns, compare job requirements, and explore remote-work opportunities.

The analysis will provide descriptive insights from the available postings. It will not measure the complete gap between employer skill demand and job-seeker skill supply because the dataset does not contain a separate dataset of job seekers' skills.

## 6. Scope and Limitations

### 6.1 Scope

The project will focus on:

- Job-posting records associated with India and the United States.
- Job roles, categories, seniority levels, and locations.
- Skill mentions in advertisements with available skills information.
- Available annual salary values recorded in USD.
- Experience requirements and remote-work labels.
- Data cleaning, exploratory analysis, visualisation, and dashboard development.
- Descriptive comparisons using Excel, SQL, Python, and Power BI.

### 6.2 Limitations

1. **Dataset coverage:** The dataset represents the job postings collected and included by its publisher, not every vacancy in India or the United States.

2. **Missing salary information:** Many postings do not contain salary values. Salary analysis will therefore use only suitable records with valid salary data, and sample sizes will be reported.

3. **Missing skills information:** Skills information is not equally complete across the two countries. Direct skill comparisons may be biased, so coverage limitations will be clearly stated.

4. **Salary interpretation:** Salary values are recorded in USD and may contain inconsistencies or unusual values. They may not represent the salary ultimately paid to a successful candidate.

5. **Country comparison:** The number of records differs between India and the United States. Raw counts alone may therefore produce misleading comparisons.

6. **Skill supply limitation:** The dataset describes job advertisements and does not independently measure the skills possessed by job seekers. Consequently, the project cannot establish a complete employer-demand-versus-job-seeker-supply gap.

7. **Descriptive analysis:** The findings will describe patterns and associations in the available data. They will not establish the causes of hiring decisions or guarantee future job-market trends.

Despite these limitations, the project aims to produce a structured and transparent analysis of the available job postings and present useful insights through an interactive dashboard.
