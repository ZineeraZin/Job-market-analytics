```python
print ("=" * 147)
print ("                                                Check outliers using IQR")
print ("=" * 147)
```

```python
# Check outliers using IQR

for col in ['experience_years', 'salary_annual_usd']:

    Q1 = df_Cleaned[col].quantile(0.25)
    Q3 = df_Cleaned[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_Cleaned[
        (df_Cleaned[col] < lower_limit) |
        (df_Cleaned[col] > upper_limit)
    ]

    print(f"\nOutliers in {col}: {len(outliers)}")
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
```

### Interpretation:

* **Experience**: 702 records were flagged because their experience values fall outside the calculated range of -5.5 to 14.5 years. Since negative experience is not realistic, those values should be checked. Higher experience values may represent genuine senior-level professionals.

* **Salary**: 526 records were flagged because their annual salaries fall outside the calculated range of -$41,400 to $168,840. Higher salaries may be genuine for senior or specialised positions, so they should not automatically be treated as errors.

# Data Cleaning – Final Summary

### The following data-cleaning steps were performed on the working copy df_Cleaned:

* 1. Removed irrelevant column: Dropped fliar_profile_url because approximately 71.62% of its values were missing and profile URLs were not required for the analysis.

* 2. Handled missing categorical values: Replaced missing values in categorical columns with "Unknown" to preserve available records.

* 3. Retained numerical missing values: Kept missing values in experience_years and salary_annual_usd because a large proportion was missing. Imputation could introduce misleading information.

* 4. Checked geographical columns: Retained city and region because location is important for job market analysis.

* 5. Duplicate check: Found zero exact duplicate records, so no rows were removed.

* 6. Data type conversion: Converted posted_date from string to datetime format to support date-based analysis.

* 7. Standardised inconsistent values: Performed the necessary consistency checks and standardisation.

* 8. Outlier investigation: Used the IQR method to identify potential outliers in experience_years and salary_annual_usd. Investigated unusual values and retained the records because there was insufficient evidence to classify them as incorrect.

* **Final result**: The cleaned dataset contains 100,009 rows and 16 columns. No rows were removed during duplicate checking or outlier treatment. Only the irrelevant fliar_profile_url column was dropped.
